```python
class Solution:
    def reverse_words(self, s):
        '''Reverse the order of words in a string
        Parameters:

        s : A string containing words 
        
        '''
        words = s.split()
        words.reverse()
        return " ".join(words)
```

```python
solution = Solution()
result = solution.reverse_words("Hello Dr George")
print (result)
```

```python

class Solution:
    def productExceptSelf(self, nums):
        """
        Return an array where each element is the product
        of all elements in nums except nums[i].

        Parameters:
            nums: A list of integers.

        """

        answer = [1] * len(nums)

        left = 1

        for i in range(len(nums)):
            answer[i] = left
            left *= nums[i]

        right = 1

        for i in range(len(nums) - 1, -1, -1):
            answer[i] *= right
            right *= nums[i]

        return answer

```

```python
solution = Solution()
nums = [1, 2, 3, 4]

result = solution.productExceptSelf(nums)

print(result)  
```
