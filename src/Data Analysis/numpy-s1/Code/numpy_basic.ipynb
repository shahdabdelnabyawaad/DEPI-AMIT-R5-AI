```python
def array_reshap(arr,row = -1,cols= -1):
    try:
        arr = np.array(arr)
        return arr.reshape((row, cols))
    except Exception as e :
        raise ValueError("Invalid Reshape operation:{e}")    
```

```python
vec = [1,2,3,4,5,6]
print ("test 1(2*3 matrix):",array_reshap(vec,2,3))
```
